```python
import numpy as np
import pandas as pd


data = pd.read_csv("house-prices.csv")

data.head()
data_dropped = data.drop("Neighborhood",axis = 1)
data_encoded = pd.get_dummies(data_dropped, columns=['Brick'], dtype=int)
print(data_encoded)


x,y = data_encoded.drop("Price", axis = 1), data_encoded["Price"]

from sklearn.model_selection import train_test_split
xtr , xt , ytr , yt = train_test_split(x,y,
                                       test_size = 0.3,
                                       random_state = 20)

from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtr,ytr)

pred = model.predict(xt)
pred_reshape = np.reshape(pred, (-1, 1))
from sklearn.metrics import r2_score

print(r2_score(pred_reshape , yt))
```
